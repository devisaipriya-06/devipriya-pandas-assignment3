#### Q1. Array operations and vectorization

**Task:**

* Predict the output of both statements.
* Identify one similarity and one difference between the two data structures.


````python
import numpy as np
import pandas as pd

# Creating a NumPy array
arr = np.array([10, 20, 30])

# Creating a Pandas Series
s = pd.Series([10, 20, 30])

# Multiplying every value in the NumPy array by 2
print(arr * 2)

# Multiplying every value in the Pandas Series by 2
print(s * 2)
````

**Answer:**

**Output:**

The NumPy array output is:

```text
[20 40 60]
```

The Pandas Series output is:

```text
0    20
1    40
2    60
dtype: int64
```

**Similarity:**
Both NumPy arrays and Pandas Series multiply each value by `2` at the same time.

**Difference:**
The NumPy array displays only the values, while the Pandas Series displays the values along with their index and data type.

**Explanation:**
When we multiply by `2`, every value in both structures is multiplied by `2`. This is called a vectorized operation because we can perform the operation on all values without writing a separate loop.

### Q1. Broadcasting
###  Two NumPy arrays have shapes (3, 1) and (1, 3). Predict the shape of their element-wise sum and explain how broadcasting works. Why might the result differ from adding two one-dimensional arrays?

**Task:**
Predict the output of the following code. Also identify the shapes of `a`, `b`, and `result`, and explain how broadcasting works.


````python
import numpy as np

# Creating a column array with 3 values
a = np.array([[10], [20], [30]])

# Creating a row array with 3 values
b = np.array([[1, 2, 3]])


result = a + b

print(result)
print(result.shape)
````

**Answer:**

The output is:

```text
[[11 12 13]
 [21 22 23]
 [31 32 33]]
(3, 3)
```

The shape of `a` is `(3, 1)` and the shape of `b` is `(1, 3)`. After broadcasting, both arrays are treated as compatible with shape `(3, 3)`, so the result also has shape `(3, 3)`.

**Explanation:**
Broadcasting allows NumPy to perform operations on arrays with different shapes by automatically expanding their dimensions when they are compatible. Here, each value in `a` is added to all three values in `b`.

If `b = np.array([1, 2])`, broadcasting would not work because the dimensions are not compatible for this operation.

### Q1. Student Assessment Data Cleaning

**Task:**
Load the student assessment dataset and inspect the first few rows, data types, and missing values.


````python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\kanch\Downloads\student_assessment_dirty.csv"
)

df.columns = df.columns.str.strip()

print(df.head())
print(df.info())
print(df.isnull().sum())
````

**Answer:**

The dataset contains **15 student assessment records** and **5 columns**.

From the inspection:

* `head()` shows the first five records.
* `info()` shows that all columns have `object` data type.
* The `Marks` column has **12 non-null values** and **3 missing values**.
* The `Marks` column is stored as `object` because it contains different types of values such as numbers, `Absent`, `NA`, and `88 marks`.

**Explanation:**
The initial inspection helps us understand the structure of the dataset and identify missing or inconsistent data before starting the cleaning process.


````python
df['Department'] = df['Department'].replace({
    'it': 'IT',
    'I.T.': 'IT',
    'finance': 'Finance'
})

print(df[['Student_ID', 'Department']])
````

**Answer:**

The department names were standardized successfully.

* `it` and `I.T.` were changed to `IT`.
* `finance` was changed to `Finance`.
* `HR` was already consistent.

**Explanation:**
Standardizing department names makes the data consistent and easier to analyze or group later.


````python
duplicates = df[df.duplicated(keep=False)]

print(duplicates)
````

````python
df = df.drop_duplicates()

print(df)
````

````python
# Remove " marks" from values such as "88 marks"
df['Marks'] = (
    df['Marks']
    .astype(str)
    .str.replace(' marks', '', regex=False)
)

# Convert Marks to numeric
df['Marks'] = pd.to_numeric(
    df['Marks'],
    errors='coerce'
)

print(df[['Student_ID', 'Student_Name', 'Marks']])
````

````python
missing_marks = df[df['Marks'].isnull()]

print(
    missing_marks[
        ['Student_ID', 'Student_Name', 'Marks']
    ]
)
````

**Answer:**

There are **4 students with missing marks** after converting the `Marks` column to numeric.

* **Meera (S105)** was marked as `Absent`.
* **Sneha (S109)** had `NA`, which should be investigated before assuming the reason.
* **Karan (S104)** and **Aman (S112)** had blank marks.

**Explanation:**
Missing marks should not be deleted immediately because they may represent different situations such as absence or marks that were not entered. Deleting them without investigation could result in loss of useful student records.


````python
invalid_marks = df[
    (df['Marks'] < 0) |
    (df['Marks'] > 100)
]

print(
    invalid_marks[
        ['Student_ID', 'Student_Name', 'Marks']
    ]
)
````

````python
# Replacing marks above 100 with NaN because they are invalid
df.loc[df['Marks'] > 100, 'Marks'] = pd.NA

# Displaying the updated DataFrame
print(df)
````

````python
# Displaying the final cleaned DataFrame
print(df)

# Reporting the cleaning results
print("\nRecords removed:", 1)
print("Missing marks remaining:", df['Marks'].isnull().sum())
print("Invalid marks identified:", 1)
````

**Answer:**

The student assessment data was cleaned successfully.

* Department names were standardized to `IT`, `HR`, and `Finance`.
* **1 exact duplicate record** was removed.
* Text values such as `88 marks` were converted to numeric form.
* Missing marks were investigated. Some missing values were due to absence or values that need further investigation.
* **1 invalid mark** was identified: Kabir's mark of `105`, which is outside the valid range of 0–100. It was changed to `NaN` while keeping the student record.
* **5 missing marks** remain in the cleaned dataset.

**Explanation:**
The cleaning process keeps useful student records while identifying duplicates, missing values, and invalid marks instead of deleting data without investigation.
